gated hops

```python
%%writefile /content/LLaMA-Factory/src/llamafactory/model/qwen2_vl_reasoning.py
import torch
import torch.nn as nn
from transformers import Qwen2VLForConditionalGeneration

class Qwen2VLReasoningModel(Qwen2VLForConditionalGeneration):
    def __init__(self, config):
        print("✅ Qwen2VLReasoningModel loaded successfully.")
        super().__init__(config)

        hidden_size = config.hidden_size  # usually 1536 for Qwen2-VL-2B

        # Stairstep reasoning layers
        self.w0 = nn.Linear(hidden_size, hidden_size)
        self.w1 = nn.Linear(hidden_size, hidden_size)

        # Optional: step-type embedding for each reasoning stage
        self.type_embed = nn.Embedding(2, hidden_size)

        # Optional classifier (not used in step 4, but can be reused later)
        self.classifier = nn.Linear(hidden_size, config.vocab_size)

    def forward(
        self,
        input_ids=None,
        attention_mask=None,
        pixel_values=None,
        image_grid_thw=None,
        labels=None,
        **kwargs
    ):
        # Step 0: Extract visual patch tokens [B, T, D]
        visual_output = self.model.visual(pixel_values, grid_thw=image_grid_thw)

        # Step 1: Apply sequential reasoning
        z1 = self.w0(visual_output)  # First reasoning stage
        z2 = self.w1(z1)             # Second reasoning stage

        # Step 2: Add time-step style embeddings
        # Expand step tokens to match batch and sequence size
        bsz, seq_len, _ = z1.size()
        z1_tagged = z1 + self.type_embed(torch.full((bsz, seq_len), 0, device=z1.device))
        z2_tagged = z2 + self.type_embed(torch.full((bsz, seq_len), 1, device=z2.device))

        # Step 3: Simulate progressive time steps by concatenating z1 → z2
        reasoned_tokens = torch.cat([z1_tagged, z2_tagged], dim=1)  # [B, 2T, D]

        # Step 4: Inject as image_features (used as prefix)
        self.model.image_features = reasoned_tokens

        # Pass through the rest of the model
        return super().forward(
            input_ids=input_ids,
            attention_mask=attention_mask,
            pixel_values=pixel_values,
            image_grid_thw=image_grid_thw,
            labels=labels,
            **kwargs
        )
```

```python
%%writefile /content/LLaMA-Factory/src/llamafactory/model/qwen2_vl_reasoning.py
import torch
import torch.nn as nn
from transformers import Qwen2VLForConditionalGeneration

class GatedReasoningLayer(nn.Module):
    def __init__(self, hidden_size):
        super().__init__()
        self.W_gate = nn.Linear(hidden_size, hidden_size)
        self.W_update = nn.Linear(hidden_size, hidden_size)
        self.activation = nn.GELU()
        self.norm = nn.LayerNorm(hidden_size)

    def forward(self, x, prev):
        gate = torch.sigmoid(self.W_gate(x))
        update = self.activation(self.W_update(x))
        return self.norm((1 - gate) * prev + gate * update)

class Qwen2VLReasoningModel(nn.Module):
    def __init__(self, config):
        super().__init__()
        self.model = Qwen2VLForConditionalGeneration(config)

        vc = config.vision_config
        if not hasattr(vc, "embed_dim"):
            print("🛠️ Patching missing embed_dim...")
            vc.embed_dim = vc.hidden_size
        if not hasattr(vc, "mlp_ratio"):
            print("🛠️ Patching missing mlp_ratio...")
            vc.mlp_ratio = 4.0

        hidden_size = vc.hidden_size
        self.num_hops = 4  # You can increase this

        self.reason_layers = nn.ModuleList([
            GatedReasoningLayer(hidden_size) for _ in range(self.num_hops)
        ])

        self.step_embed = nn.Embedding(self.num_hops, hidden_size)

    def forward(
        self,
        input_ids=None,
        attention_mask=None,
        pixel_values=None,
        image_grid_thw=None,
        labels=None,
        **kwargs
    ):
        visual_output = self.model.model.visual(pixel_values, grid_thw=image_grid_thw)

        if visual_output.dim() == 2:
            visual_output = visual_output.unsqueeze(0)
        elif visual_output.shape[0] != pixel_values.shape[0]:
            visual_output = visual_output.permute(1, 0, 2)

        if visual_output.dim() != 3:
            raise ValueError(f"❌ visual_output has invalid shape: {visual_output.shape}")

        B, T, D = visual_output.shape
        device = visual_output.device

        z = visual_output
        outputs = []

        for i, layer in enumerate(self.reason_layers):
            step_embed = self.step_embed(torch.full((B, T), i, device=device))
            z = layer(z + step_embed, visual_output)
            outputs.append(z)

        reasoned_tokens = torch.cat(outputs, dim=1)  # [B, T * hops, D]

        self.model.model.image_features = reasoned_tokens.view(-1, D)  # [B*T*hops, D]
        self.model.model.image_mask = torch.ones(B, T * self.num_hops, dtype=torch.bool, device=device)

        assert self.model.model.image_features.size(0) == self.model.model.image_mask.sum().item(), (
            f"Mismatch: image_features ({self.model.model.image_features.size(0)}) != image_mask sum ({self.model.model.image_mask.sum().item()})"
        )

        print(f"✅ Injected image_features: {self.model.model.image_features.shape}, image_mask: {self.model.model.image_mask.shape}")

        return self.model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            pixel_values=pixel_values,
            image_grid_thw=image_grid_thw,
            labels=labels,
            **kwargs
        )
```

```python
%%writefile /content/LLaMA-Factory/src/llamafactory/model/qwen2_vl_reasoning.py
import torch
import torch.nn as nn
from transformers import Qwen2VLForConditionalGeneration

class ReasoningSelfAttentionLayer(nn.Module):
    def __init__(self, hidden_size, num_heads=8):
        super().__init__()
        print(f"🔧 Initializing ReasoningSelfAttentionLayer with hidden_size={hidden_size}, num_heads={num_heads}")
        self.self_attn = nn.MultiheadAttention(embed_dim=hidden_size, num_heads=num_heads, batch_first=True)
        self.ffn = nn.Sequential(
            nn.Linear(hidden_size, hidden_size * 4),
            nn.GELU(),
            nn.Linear(hidden_size * 4, hidden_size),
        )
        self.norm1 = nn.LayerNorm(hidden_size)
        self.norm2 = nn.LayerNorm(hidden_size)

    def forward(self, x):
        attn_output, _ = self.self_attn(x, x, x)
        x = self.norm1(x + attn_output)
        x = self.norm2(x + self.ffn(x))
        return x

class Qwen2VLReasoningModel(Qwen2VLForConditionalGeneration):
    def __init__(self, config):
        print("🚀 Initializing Qwen2VLReasoningModel...")
        super().__init__(config)

        vc = config.vision_config
        if not hasattr(vc, "embed_dim"):
            print("🛠️ Patching missing embed_dim...")
            vc.embed_dim = vc.hidden_size
        if not hasattr(vc, "mlp_ratio"):
            print("🛠️ Patching missing mlp_ratio...")
            vc.mlp_ratio = 4.0

        hidden_size = vc.hidden_size
        print(f"✅ Vision hidden size detected: {hidden_size}")

        self.num_hops = 4
        print(f"🔁 Initializing {self.num_hops} reasoning layers...")

        self.reason_layers = nn.ModuleList([
            ReasoningSelfAttentionLayer(hidden_size) for _ in range(self.num_hops)
        ])

        # Make sure reasoning layers are trainable
        for param in self.reason_layers.parameters():
            param.requires_grad = True

        print("✅ Reasoning layers initialized and set as trainable.")

        # Optional: show trainable reasoning parameters
        print("📋 Reasoning layer parameters:")
        for name, param in self.named_parameters():
            if "reason_layers" in name and param.requires_grad:
                print("  🧠", name)

    def forward(
        self,
        input_ids=None,
        attention_mask=None,
        pixel_values=None,
        image_grid_thw=None,
        labels=None,
        **kwargs
    ):
      # Only run reasoning if pixel_values are provided
        if pixel_values is not None:
            visual_output = self.model.visual(pixel_values, grid_thw=image_grid_thw)

            if visual_output.dim() == 2:
                visual_output = visual_output.unsqueeze(0)
            elif visual_output.shape[0] != pixel_values.shape[0]:
                visual_output = visual_output.permute(1, 0, 2)

            if visual_output.dim() != 3:
                raise ValueError(f"❌ visual_output has invalid shape: {visual_output.shape}")

            B, T, D = visual_output.shape
            device = visual_output.device

            z = visual_output
            outputs = []

            for i, layer in enumerate(self.reason_layers):
                z = layer(z)
                outputs.append(z)

            reasoned_tokens = torch.cat(outputs, dim=1)  # [B, T*num_hops, D]
            self.model.image_features = reasoned_tokens.view(-1, D)
            self.model.image_mask = torch.ones(B, T * self.num_hops, dtype=torch.bool, device=device)

        assert self.model.image_features.size(0) == self.model.image_mask.sum().item(), (
            f"Mismatch: image_features ({self.model.image_features.size(0)}) != image_mask sum ({self.model.image_mask.sum().item()})"
        )

        #print(f"✅ Injected image_features: {self.model.image_features.shape}, image_mask: {self.model.image_mask.shape}")

        return super().forward(
            input_ids=input_ids,
            attention_mask=attention_mask,
            pixel_values=pixel_values,
            image_grid_thw=image_grid_thw,
            labels=labels,
            **kwargs
        )
```
