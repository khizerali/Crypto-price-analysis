```python
import json

def convert_to_vllm_format(input_path, output_path):
    with open(input_path, "r") as f:
        original_data = json.load(f)

    new_data = []

    for i, item in enumerate(original_data):
        conversations = item.get("conversations", [])
        images = item.get("images")
        if not images or len(images) == 0:
            print(f"⚠️ Skipping item {i} due to missing or empty 'images' field.")
            continue

        image_path = images[0]
        messages = []
        image_refs = []

        for j, conv in enumerate(conversations):
            role = "user" if conv["from"] == "human" else "assistant"
            text = conv["value"]

            # Only insert <image> for user messages (optional: first only, or each?)
            if role == "user":
                if "<image>" not in text:
                    if j == 0:
                        text = f"<image>{text}"
                    else:
                        text = f"{text}<image>"
                image_refs.append(image_path)

            messages.append({
                "role": role,
                "content": text
            })

        new_data.append({
            "messages": messages,
            "images": image_refs if image_refs else [image_path]
        })

    with open(output_path, "w") as f:
        json.dump(new_data, f, indent=2)

    print(f"✅ Conversion complete. Output written to: {output_path}")

# Example usage:
input_json_path = "/content/data.json"
output_json_path = "/content/LLaMA-Factory/data/plastic_dataset_new.json"
convert_to_vllm_format(input_json_path, output_json_path)
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from PIL import Image
import requests
from io import BytesIO

# Load the merged model and tokenizer
model_path = "/content/qwen2vl_2b_instruct_lora_merged"  # Path to your merged model
tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    model_path,
    device_map="auto",
    trust_remote_code=True
).eval()

# Function to process text-only queries
def text_inference(query):
    inputs = tokenizer(query, return_tensors='pt').to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=512)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response

# Function to process image-text queries
def visual_inference(image_path, query):
    # Load image
    if image_path.startswith('http'):
        response = requests.get(image_path)
        image = Image.open(BytesIO(response.content)).convert('RGB')
    else:
        image = Image.open(image_path).convert('RGB')

    # Process query
    query = tokenizer.from_list_format([
        {'image': image_path},  # Either path or URL
        {'text': query},
    ])

    inputs = tokenizer(query, return_tensors='pt').to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=512)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response

# Example usage
if __name__ == "__main__":
    # Text-only example
    text_query = "Tell me about the capabilities of this AI model."
    print("Text-only response:")
    print(text_inference(text_query))
    print("\n" + "="*50 + "\n")

    # Image-text example (replace with your image path/URL)
    image_url = "/content/LLaMA-Factory/data/images/001077.png"  # Replace with actual image
    visual_query = "What is shown in this image?"
    print("Visual response:")
    print(visual_inference(image_url, visual_query))
```

```python
def process_image_text_query_manual(image_path, prompt):
    from PIL import Image
    import requests
    from io import BytesIO
    import os

    def load_image(image_path):
        if image_path.startswith(("http://", "https://")):
            response = requests.get(image_path)
            image = Image.open(BytesIO(response.content)).convert("RGB")
        else:
            if not os.path.exists(image_path):
                raise FileNotFoundError(f"Image not found: {image_path}")
            image = Image.open(image_path).convert("RGB")
        return image

    try:
        image = load_image(image_path)

        # Tokenize the prompt manually
        encoded = tokenizer(prompt, return_tensors="pt", add_special_tokens=True)
        encoded = {k: v.to(model.device) for k, v in encoded.items()}

        # Process image manually
        image_inputs = processor.image_processor(images=image, return_tensors="pt").to(model.device)

        # Create input dict
        inputs = {
            "input_ids": encoded["input_ids"],
            "attention_mask": encoded["attention_mask"],
            "images": image_inputs["pixel_values"]
        }

        outputs = model.generate(**inputs, max_new_tokens=512)
        return tokenizer.decode(outputs[0], skip_special_tokens=True)

    except Exception as e:
        return f"❌ Manual inference error: {e}"
```
