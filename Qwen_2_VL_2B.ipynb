```python
import os
import shutil

# Source and destination directories
source_dir = "/content/LLaMA-Factory/train_data/images/test"
dest_dir = "/content/LLaMA-Factory/data/plastic_data/train_data/images/"

# Ensure the destination directory exists
os.makedirs(dest_dir, exist_ok=True)

# Loop through all files in the source directory
for filename in os.listdir(source_dir):
    # Construct the full file path
    source_file = os.path.join(source_dir, filename)
    dest_file = os.path.join(dest_dir, filename)

    # Check if it's a file (not a directory)
    if os.path.isfile(source_file):
        # Copy the file to the destination
        shutil.copy(source_file, dest_file)
        #print(f"Copied: {filename}")

```

```python
import json
import os

def remove_extra_images(json_file, image_folder):
    # Step 1: Load the JSON data
    with open(json_file, 'r') as f:
        data = json.load(f)

    # Step 2: Create a set of all images that are mentioned in the JSON data
    images_in_json = set()

    for entry in data:
        # Each entry in the JSON contains a "messages" field that has the <image> tokens
        for message in entry.get("messages", []):
            if "content" in message:
                content = message["content"]
                # Count the <image> tokens and collect the corresponding image paths
                images_in_json.update([img for img in content.split() if img.startswith('<image>')])

    # Step 3: Extract image paths from the content of the <image> tokens in the JSON
    image_paths_in_json = set()
    for image_token in images_in_json:
        # Extract the actual image paths from <image> tags, e.g., <image>images/001075.png</image>
        image_path = image_token.replace("<image>", "").replace("</image>", "").strip()
        image_paths_in_json.add(image_path)

    # Step 4: Get all the images in the folder
    all_images_in_folder = set(os.listdir(image_folder))

    # Step 5: Identify extra images (those in the folder but not in the JSON)
    extra_images = all_images_in_folder - image_paths_in_json

    # Step 6: Remove the extra images from the folder
    for extra_image in extra_images:
        image_path = os.path.join(image_folder, extra_image)
        try:
            os.remove(image_path)
            print(f"Removed extra image: {extra_image}")
        except Exception as e:
            print(f"Error removing image {extra_image}: {e}")

# Example Usage
json_file = "/content/LLaMA-Factory/LLaMA-Factory/data/plastic_dataset/data.json"  # Provide the path to your JSON file
image_folder = "/content/LLaMA-Factory/LLaMA-Factory/train_data/images"  # Provide the path to your image folder

remove_extra_images(json_file, image_folder)
```
